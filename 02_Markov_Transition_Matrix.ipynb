# Building Markov Transition Matrix from input sequence

```python
transition_matrix = {'AA': 0, 'AC': 0, 'AG': 0, 'AT': 0,
              'CA': 0, 'CC': 0, 'CG': 0, 'CT': 0,
              'GA': 0, 'GC': 0, 'GG': 0, 'GT': 0,
              'TA': 0, 'TC': 0, 'TG': 0, 'TT': 0}
total_count = {'A' : 0, 'C': 0, 'G': 0, 'T': 0}
count_map = {'AA': 0, 'AC': 0, 'AG': 0, 'AT': 0,
              'CA': 0, 'CC': 0, 'CG': 0, 'CT': 0,
              'GA': 0, 'GC': 0, 'GG': 0, 'GT': 0,
              'TA': 0, 'TC': 0, 'TG': 0, 'TT': 0}
total_count = {'A' : 0, 'C': 0, 'G': 0, 'T': 0}
s = input("Enter a string: ")
for i in range(len(s) - 1):
    count_map[s[i] + s[i + 1]] += 1
    total_count[s[i]] += 1
for i in count_map.keys():
    transition_matrix[i] = count_map[i] / total_count[i[0]]
print(transition_matrix)

```
