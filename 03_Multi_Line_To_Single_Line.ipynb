```python
input_file = 'multiline_input.fasta'
output_file = 'singleline_output.fasta'
```

```python
with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
    first_record = True
    for line in infile:
        line = line.rstrip('\n')
        if line.startswith('>'):
            if not first_record:
                outfile.write('\n') 
            outfile.write(line + '\n')
            first_record = False
        else:
            outfile.write(line)
    outfile.write('\n')

    #This implementation ensures pure streaming solution and no list is used to store the sequences.
```

```python
print('Converted FASTA file to single-line format:')
with open(output_file) as f:
    print(f.read())
```
